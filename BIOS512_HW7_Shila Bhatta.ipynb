## Question 2
#### Explain how patches relate to the idea of committing a change to the history of your code?

````python
A patch is a snapshot of changes that we make in the code in a file like addition, modification, removal. 
After the changes are made, these changes are then committed to the repository.
Commit records the metadata such as the history including who and what changes were made.
````

````python
Branch is the part of the original repositpry used to work on changes independently within a same project while 
a fork is the clone or copy of the entirely other repository allowing others to modify and contribute without changing original.
````
